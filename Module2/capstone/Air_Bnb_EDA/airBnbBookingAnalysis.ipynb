## Data Cleaning

### Handle Missing Values

````python
# Identify missing data and decide on an approach to handle it (imputation, removal, etc.).

df.isna().sum()
````

````python
# Impute missing values for 'name' and 'host_name'

df['name'] = df['name'].fillna('Unknown')
df['host_name'] = df['host_name'].fillna('Unknown')
````

````python
# checking the last review & reviews_per_month have missing values in common rows

df[df['last_review'].isna().eq(True) & df['reviews_per_month'].isna().eq(True)].shape
````

````python
# checking if the review & reviews_per_month have missing values beacuase of 0 number_of_reviews.

df[df['last_review'].isna().eq(True) & df['reviews_per_month'].isna().eq(True) & df['number_of_reviews'].eq(0)].shape
````

> Note: missing values of last_review & reviews_per_month are due to number_of_reviews are 0 so imputing them with ***0*** and ***no review*** respectively

````python
# Impute missing values for 'last_review' and 'reviews_per_month'

df['last_review'] = df['last_review'].fillna('No Review')
df['reviews_per_month'] = df['reviews_per_month'].fillna(0)
````

````python
# Check if there are still missing values
print(df.isna().sum())
````

### Duplicate Check

````python
# Duplicate values in the dataset 

df.duplicated().sum()
````

````python
df.info()
````

### Outliers

````python
df.describe()
````

````python
plt.figure(figsize=(10, 5))
sns.boxplot(data=df['price'])
plt.show()
````

````python
# Calculate IQR for price
Q1 = df['price'].quantile(0.25)
Q3 = df['price'].quantile(0.75)
IQR = Q3 - Q1

# filtering outliers
df = df[(df['price'] >= (Q1 - 1.5 * IQR)) & (df['price'] <= (Q3 + 1.5 * IQR))]
df.shape
````

````python
sns.boxplot(data=df['price'])
plt.show()
````

## Feature Engineering

### Year Extraction & Enrichment

````python
### Extracting Year from last review and saving it in new column last review year

df['last_review_year'] = pd.to_datetime(df['last_review'], errors='coerce').dropna().dt.year.astype('Int32')
df['last_review_month'] = pd.to_datetime(df['last_review'], errors='coerce').dropna().dt.month.astype('Int32')

df.sample()
````

#### 1. How are Airbnb listings distributed across different neighborhood groups in NYC?

````python
# 1. How are Airbnb listings distributed across different neighborhoods in NYC?

airbnb_listing = df.groupby(by='neighbourhood_group').size().reset_index(name="Number of listings")
airbnb_listing
````

#### Chart - 1

````python
# chart - 1 : distribution of airbnb listings across neighborhoods in NYC

plt.figure(figsize=(10,6))

x = 'neighbourhood_group'
y = 'Number of listings'
data = airbnb_listing

# Bar plot
ax = sns.barplot(x = x, y = y, data=data, hue = x, palette='viridis')


# Add text labels on top of the bars
for i in ax.containers:
    ax.bar_label(i)

# Add Title and labels to the axis
plt.title('Distribution of Airbnb Listings Across neighborhood Groups in NYC')
plt.xlabel('Neighborhood Groups')
plt.ylabel('Number of Listings')


plt.show()
````

##### 1. Why did you pick the specific chart?

**Answer Here**


I picked a bar chart because it effectively visualizes categorical data, which in this case represents different neighborhoods and their number of Airbnb listings. Bar charts are ideal for comparing the size of each category, making it easy to see how listings are distributed across various neighborhoods. This allows for a clear comparison of each neighbourhood’s market presence at a glance.

##### 2. What is/are the insight(s) found from the chart?

**Answer Here**

The bar chart provides clear insights into the distribution of Airbnb listings across NYC:

- **Manhattan** leads with the highest number of listings (19,506), marking it as a major hub for short-term rentals.
- **Brooklyn** is also prominent with a high count of listings (19,415), reflecting its strong presence in the Airbnb market.
- **Queens** shows a moderate number of listings (5,667), suggesting it's a significant but less concentrated market compared to Manhattan and Brooklyn.
- **Bronx** and **Staten Island** have relatively few listings (1,070 and 365 respectively), indicating they are less active in the Airbnb market.

#### 2. What is the average price of listings by room type and neighborhood?

````python
# 2. What is the average price of listings by room type and neighborhood?

avg_price = df.groupby(['neighbourhood_group', 'room_type'])['price'].mean().reset_index()
````

#### Chart - 2

````python

# chart - 2 : Average price by room type and neighborhood group

plt.figure(figsize=(10, 6))


x = 'neighbourhood_group'
y = 'price'
    
data = avg_price

ax = sns.barplot(x = x, y = y, data = data, hue = 'room_type',  palette='viridis')
    
for i in ax.containers:
    ax.bar_label(i, fmt='%.2f')

plt.title('Average Price by Room Type and Neighborhood Group')
plt.xlabel('Neigbhorhood Groups')
plt.ylabel('Average Price')

plt.legend(title='Room Type')

plt.show()
````

A bar chart clearly displays the average prices across different neighborhoods and room types, making it easy to compare.
