## Data Cleaning

### Handle Missing Values

````python
# Identify missing data and decide on an approach to handle it (imputation, removal, etc.).

df.isna().sum()
````

````python
# Impute missing values for 'name' and 'host_name'

df['name'] = df['name'].fillna('Unknown')
df['host_name'] = df['host_name'].fillna('Unknown')
````

````python
# checking the last review & reviews_per_month have missing values in common rows

df[df['last_review'].isna().eq(True) & df['reviews_per_month'].isna().eq(True)].shape
````

````python
# checking if the review & reviews_per_month have missing values beacuase of 0 number_of_reviews.

df[df['last_review'].isna().eq(True) & df['reviews_per_month'].isna().eq(True) & df['number_of_reviews'].eq(0)].shape
````

> Note: missing values of last_review & reviews_per_month are due to number_of_reviews are 0 so imputing them with ***0*** and ***no review*** respectively

````python
# Impute missing values for 'last_review' and 'reviews_per_month'

df['last_review'] = df['last_review'].fillna('No Review')
df['reviews_per_month'] = df['reviews_per_month'].fillna(0)
````

````python
# Check if there are still missing values
print(df.isna().sum())
````

````python
# chart - 1 : distribution of airbnb listings across neighborhoods in NYC

plt.figure(figsize=(10,6))

x = 'neighbourhood_group'
y = 'Number of listings'
data = airbnb_listing

# Bar plot
ax = sns.barplot(x = x, y = y, data=data, hue = x, palette='viridis')


# Add text labels on top of the bars
for i in ax.containers:
    ax.bar_label(i)

# Add Title and labels to the axis
plt.title('Distribution of Airbnb Listings Across neighborhood Groups in NYC')
plt.xlabel('Neighborhood Groups')
plt.ylabel('Number of Listings')


plt.show()
````

#### 12. How does the availability of listings vary across different neighborhoods and room types?

````python

# Group by neighborhood and room type, then calculate the average availability
availability_data = df.groupby(['neighbourhood_group', 'room_type'])['availability_365'].mean().reset_index()
````

````python
# Plotting
plt.figure(figsize=(14, 8))
ax = sns.barplot(data=availability_data, x='neighbourhood_group', y='availability_365', hue='room_type', palette='viridis')

for i in ax.containers:
    ax.bar_label(i, fmt='%.2f')

plt.title('Average Availability of Listings Across Neighborhoods and Room Types')
plt.xlabel('Neighborhood Group')
plt.ylabel('Average Availability (days)')
plt.legend(title='Room Type')
plt.show()
````

#### 13. What is the impact of minimum night stay requirements on booking rates in different neighborhoods?

````python
# Group by neighborhood and minimum nights, then calculate the average booking rate (using reviews_per_month as a proxy)
min_nights_booking = df.groupby(['neighbourhood_group', 'minimum_nights'])['reviews_per_month'].mean().reset_index()
````

````python
# Plotting
plt.figure(figsize=(14, 8))
sns.histplot(data=min_nights_booking, x='minimum_nights', y='reviews_per_month', hue='neighbourhood_group')
plt.title('Impact of Minimum Night Stay Requirements on Booking Rates Across Neighborhoods')
plt.xlabel('Minimum Nights')
plt.ylabel('Average Booking Rate (reviews per month)')
plt.legend(title='Neighborhood Group')
plt.show()
````

#### 14. What are the seasonal trends in availability and booking for different neighborhoods?


````python
# Group by neighborhood and review month, then calculate the average availability and booking rates
seasonal_data = df.groupby(['neighbourhood_group', 'last_review_month']).agg(
    avg_availability=('availability_365', 'mean'),
    avg_reviews_per_month=('reviews_per_month', 'mean')
).reset_index()

# Create subplots for availability and booking trends
fig, axes = plt.subplots(nrows=1, ncols=2, figsize=(20, 8))

# Plotting availability trends
sns.lineplot(data=seasonal_data, x='last_review_month', y='avg_availability', hue='neighbourhood_group', marker='o', palette='viridis', ax=axes[0])
axes[0].set_title('Seasonal Trends in Availability Across Neighborhoods')
axes[0].set_xlabel('Month')
axes[0].set_ylabel('Average Availability (days)')
axes[0].legend(title='Neighborhood Group')

# Plotting booking trends
sns.lineplot(data=seasonal_data, x='last_review_month', y='avg_reviews_per_month', hue='neighbourhood_group', marker='o', palette='viridis', ax=axes[1])
axes[1].set_title('Seasonal Trends in Booking Rates Across Neighborhoods')
axes[1].set_xlabel('Month')
axes[1].set_ylabel('Average Booking Rate (reviews per month)')
axes[1].legend(title='Neighborhood Group')

# Adjust layout
plt.tight_layout()
plt.show()
````

#### 15. What neighborhoods show the highest potential for new listings based on current demand and pricing trends?

````python
neighborhood_potential_sorted = neighborhood_potential.sort_values(by=['avg_availability', 'avg_price'], ascending=[False, True])
````

````python
plt.figure(figsize=(14, 8))
ax = sns.barplot(data=neighborhood_potential_sorted, x='neighbourhood_group', y='avg_price', hue='avg_availability', palette='viridis')

for i in ax.containers:
    ax.bar_label(i, fmt='%.2f')

plt.title('Potential for New Listings Based on Pricing Trends and Availability')
plt.xlabel('Neighborhood')
plt.ylabel('Average Price')

plt.legend(title='Average Availability', labels=[])
plt.show()
````

#### 16. How does the listing price impact the number of bookings and the average number of reviews in different neighborhoods?

````python
df['price_range'] = pd.cut(df['price'], bins=[0, 100, 200, 300, 400, 500], labels=['0-100', '100-200', '200-300', '300-400', '400-500'])
price_impact = df.groupby('price_range').agg(
    avg_reviews_per_month=('reviews_per_month', 'mean'),
    avg_number_of_reviews=('number_of_reviews', 'mean')
).reset_index()
````

````python
fig, axes = plt.subplots(nrows=1, ncols=2, figsize=(20, 8))

sns.barplot(data=price_impact, x='price_range', y='avg_reviews_per_month', ax=axes[0], hue='price_range', palette='viridis')
axes[0].set_title('Average Reviews per Month by Price Range')
axes[0].set_xlabel('Price Range')
axes[0].set_ylabel('Average Reviews per Month')

sns.barplot(data=price_impact, x='price_range', y='avg_number_of_reviews', ax=axes[1], hue='price_range', palette='viridis')
axes[1].set_title('Average Number of Reviews by Price Range')
axes[1].set_xlabel('Price Range')
axes[1].set_ylabel('Average Number of Reviews')

plt.tight_layout()
plt.show()
````

#### 17. Which neighborhoods have the most potential for increasing prices based on the correlation between price and number of reviews?

````python
price_reviews_corr = df.groupby('neighbourhood_group').apply(
    lambda x: x[['price', 'number_of_reviews']].corr().iloc[0, 1]
).reset_index(name='price_reviews_correlation')
````

````python
potential_neighborhoods = price_reviews_corr.sort_values(by='price_reviews_correlation', ascending=False)
potential_neighborhoods
````

#### 19. What is the impact of listing title length on the number of bookings and reviews?
